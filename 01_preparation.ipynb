```python
from pathlib import Path

import pandas as pd

ROOT = Path.cwd()
RAW = ROOT / "data" / "raw"
PROCESSED = ROOT / "data" / "processed"

PROCESSED.mkdir(parents=True, exist_ok=True)

print("Dossier de travail :", ROOT)

fichiers_xpt = [
    fichier
    for fichier in RAW.rglob("*")
    if fichier.suffix.lower() == ".xpt"
]

print("Fichiers XPT trouvés :", fichiers_xpt)

if len(fichiers_xpt) != 1:
    raise ValueError(
        "Place un seul fichier XPT BRFSS 2024 dans data/raw."
    )

source = fichiers_xpt[0]
```

```python
df = pd.read_sas(
    source,
    format="xport",
    encoding="latin-1"
)

print(f"Nombre de lignes : {df.shape[0]:,}")
print(f"Nombre de colonnes : {df.shape[1]:,}")

display(df.head())
```

```python
df.info()

memoire_mo = df.memory_usage(deep=True).sum() / 1024**2
print(f"Mémoire occupée : {memoire_mo:.1f} Mo")
```

```python
print(df.columns.tolist())
```

```python
resume = pd.DataFrame({
    "type": df.dtypes.astype(str),
    "valeurs_manquantes": df.isna().sum(),
    "pourcentage_manquant": df.isna().mean() * 100,
})

display(
    resume.sort_values(
        "pourcentage_manquant",
        ascending=False
    ).head(30)
)
```

```python
destination = PROCESSED / "brfss_2024_brut.parquet"

df.to_parquet(
    destination,
    engine="pyarrow",
    compression="snappy",
    index=False
)

print("Fichier créé :", destination)
print(f"Taille : {destination.stat().st_size / 1024**2:.1f} Mo")
```

```python
apercu = pd.read_parquet(
    destination,
    columns=df.columns[:5].tolist()
)

assert len(apercu) == len(df)

display(apercu.head())
```
